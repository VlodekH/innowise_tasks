**8. Among whom is the proportion of high-earning individuals (>50K) higher: married or single men (*marital-status* indicator)? We consider those whose *marital-status* begins with *Married* (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse) to be married; the rest are considered single.**

```python
marriage_status = ['Married-civ-spouse', 'Married-spouse-absent', 'Married-AF-spouse']

men = data[data['sex'] == 'Male']
married_men = men[men['marital-status'].isin(marriage_status)]
single_men = men[~men['marital-status'].isin(marriage_status)]

married_proportion = (married_men['salary'] == '>50K').mean()
single_proportion = (single_men['salary'] == '>50K').mean()

print(married_proportion)
print(single_proportion)
```

p**9. What is the maximum number of hours a person works per week (hours-per-week indicator)? How many people work that many hours, and what percentage of them earn a lot?**

```python
max_hours = data['hours-per-week'].max()
print(f'Max hours: {max_hours}')
npeople_max_hours = data[data['hours-per-week'] == max_hours]
print(f'Number of people working max hours: {npeople_max_hours.shape[0]}')
earn_a_lot_prec = npeople_max_hours[npeople_max_hours['salary'] == '>50K'].shape[0] / npeople_max_hours.shape[0]
print(f'Percentage of people earning >50K among those working max hours: {earn_a_lot_prec * 100:.2f}%')
```

**11.Group people into age groups *young*, *adult*, *retiree*, where:**
* *young* corresponds to 16-35 years old
* *adult* - 35-70 years old
* *retiree* - 70-100 years old

**Enter the name of the appropriate age group for each person in the new AgeGroup column.**

```python
age_groups = pd.cut(data['age'], bins = [16, 35, 70 ,100], labels = ['young', 'adult', 'retiree'], include_lowest = True)
data['AgeGroup'] = age_groups
data.head()
```

**12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people most often earn more than 50K (>50K)**

```python
earnings_by_age_group = data.groupby('AgeGroup')['salary'].apply(lambda x: (x == '>50K').sum())
print(earnings_by_age_group)

most_earning_age_group = earnings_by_age_group.idxmax()
print(f"The age group in which people most often earn more than 50K is: {most_earning_age_group}")
```

**14. Group people by employment type (the occupation column) and determine the number of people in each group. Then write a filter function, filter_func, that will return only those groups in which the average age (the age column) is 40 or less and in which all employees work more than 5 hours per week (the hours-per-week column).**

```python
data.groupby('occupation').size()
```

```python
def filter_func(data):
    grouped = data.groupby('occupation')
    filtered = grouped.filter(lambda x: x['age'].mean() <= 40 and (x['hours-per-week'] > 5).all())
    return filtered.groupby('occupation').size()
filter_func(data)
```
