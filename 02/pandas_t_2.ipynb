```python
def prepare_annual_data(df: pd.DataFrame):
    df = df.copy()

    # produced_material_quantity повторяется для каждого компонента,
    # поэтому сначала схлопываем material внутри месяца.

    monthly_materials = (
        df.groupby(
            [
                "plant_id",
                "year",
                "month",
                "produced_material",
                "produced_material_release_type",
                "produced_material_production_type",
            ],
            dropna=False,
            as_index=False,
        )
        .agg(
            produced_material_quantity=(
                "produced_material_quantity",
                "first"
            )
        )
    )

    # Теперь можно безопасно суммировать месяцы -> год
    annual_materials = (
        monthly_materials.groupby(
            [
                "plant_id",
                "year",
                "produced_material",
                "produced_material_release_type",
                "produced_material_production_type",
            ],
            dropna=False,
            as_index=False,
        )
        .agg(
            produced_material_quantity=(
                "produced_material_quantity",
                "sum"
            )
        )
    )

    # Здесь component quantity относится именно к конкретной связи,
    # поэтому ее уже можно складывать между месяцами.

    annual_edges = (
        df.groupby(
            [
                "plant_id",
                "year",
                "produced_material",
                "produced_material_release_type",
                "produced_material_production_type",
                "component_material",
                "component_material_release_type",
                "component_material_production_type",
            ],
            dropna=False,
            as_index=False,
        )
        .agg(
            component_material_quantity=(
                "component_material_quantity",
                "sum"
            )
        )
    )

    # Добавляем годовое production quantity текущего material
    annual_edges = annual_edges.merge(
        annual_materials[
            [
                "plant_id",
                "year",
                "produced_material",
                "produced_material_quantity",
            ]
        ],
        on=[
            "plant_id",
            "year",
            "produced_material",
        ],
        how="left",
    )

    return annual_materials, annual_edges
```

```python
def build_lookups(annual_materials, annual_edges):

    material_lookup = {}

    for _, row in annual_materials.iterrows():

        key = (
            row["plant_id"],
            row["year"],
            row["produced_material"],
        )

        material_lookup[key] = row.to_dict()

    edges_lookup = {}

    for _, row in annual_edges.iterrows():

        key = (
            row["plant_id"],
            row["year"],
            row["produced_material"],
        )

        edges_lookup.setdefault(key, []).append(row.to_dict())

    return material_lookup, edges_lookup
```

```python
def walk_bom(
    plant_id,
    year,
    current_material,
    fin_data,
    material_lookup,
    edges_lookup,
    result,
    path=None,
):
    if path is None:
        path = set()

    current_key = (
        plant_id,
        year,
        current_material,
    )

    # Защита от циклов типа A -> B -> C -> A
    if current_key in path:
        print(f"Cycle detected: {current_key}")
        return

    path = path | {current_key}

    # Есть ли у текущего материала компоненты?
    edges = edges_lookup.get(current_key, [])

    if not edges:
        return

    current_material_data = material_lookup.get(current_key)

    if current_material_data is None:
        return

    for edge in edges:

        component = edge["component_material"]

        result.append(
            {
                "plant": plant_id,

                "fin_material_id":
                    fin_data["produced_material"],

                "fin_material_release_type":
                    fin_data["produced_material_release_type"],

                "fin_material_production_type":
                    fin_data["produced_material_production_type"],

                "fin_production_quantity":
                    fin_data["produced_material_quantity"],

                "prod_material_id":
                    current_material,

                "prod_material_release_type":
                    current_material_data[
                        "produced_material_release_type"
                    ],

                "prod_material_production_type":
                    current_material_data[
                        "produced_material_production_type"
                    ],

                "prod_material_production_quantity":
                    current_material_data[
                        "produced_material_quantity"
                    ],

                "component_id":
                    component,

                "component_material_release_type":
                    edge["component_material_release_type"],

                "component_material_production_type":
                    edge["component_material_production_type"],

                "component_consumption_quantity":
                    edge["component_material_quantity"],

                "year":
                    year,
            }
        )

        # Главное место всей задачи:
        #
        # component текущей строки
        # становится produced_material следующего уровня.

        next_key = (
            plant_id,
            year,
            component,
        )

        if next_key in edges_lookup:

            walk_bom(
                plant_id=plant_id,
                year=year,
                current_material=component,
                fin_data=fin_data,
                material_lookup=material_lookup,
                edges_lookup=edges_lookup,
                result=result,
                path=path,
            )
```

```python
def explode_all_fin(
    annual_materials,
    annual_edges,
):
    material_lookup, edges_lookup = build_lookups(
        annual_materials,
        annual_edges,
    )

    result = []

    fin_materials = annual_materials[
        annual_materials[
            "produced_material_release_type"
        ].eq("FIN")
    ]

    for _, fin_row in fin_materials.iterrows():

        plant_id = fin_row["plant_id"]
        year = fin_row["year"]
        fin_material = fin_row["produced_material"]

        fin_key = (
            plant_id,
            year,
            fin_material,
        )

        # FIN -> первый production material
        fin_edges = edges_lookup.get(fin_key, [])

        for fin_edge in fin_edges:

            first_prod_material = (
                fin_edge["component_material"]
            )

            first_prod_key = (
                plant_id,
                year,
                first_prod_material,
            )

            # Если component FIN действительно производится,
            # начинаем explosion.
            if first_prod_key in material_lookup:

                walk_bom(
                    plant_id=plant_id,
                    year=year,
                    current_material=first_prod_material,
                    fin_data=fin_row.to_dict(),
                    material_lookup=material_lookup,
                    edges_lookup=edges_lookup,
                    result=result,
                )

    result_df = pd.DataFrame(result)

    return result_df
```

```python
annual_materials, annual_edges = prepare_annual_data(df)

bom_exploded = explode_all_fin(
    annual_materials,
    annual_edges,
)

display(bom_exploded.tail(10))

print(bom_exploded.shape)
print(bom_exploded.columns.tolist())
```
